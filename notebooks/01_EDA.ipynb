# Basic transformations

```python
def preprocess_df(df: pd.DataFrame) -> pd.DataFrame:
    """
    Convert the 't' column to Europe/Paris timezone, parse WKB coordinates into latitude and longitude,
    and remove raw geometry columns.
    """
    df_processed = df.copy()
    # 1) Convert timestamp to Europe/Paris timezone
    df_processed["t_paris"] = df_processed["t"].dt.tz_convert("Europe/Paris")

    # add date column for easier filtering
    df_processed["date"] = df_processed["t_paris"].dt.strftime("%Y-%m-%d")

    # 2) Parse WKB bytes into Shapely geometry
    df_processed["geometry"] = df_processed["coordonnees_geo"].apply(wkb.loads)

    # 3) Extract longitude (x) and latitude (y)
    df_processed["lon"] = df_processed["geometry"].apply(lambda point: point.x)
    df_processed["lat"] = df_processed["geometry"].apply(lambda point: point.y)

    # 4) Drop raw geometry columns
    df_processed.drop(columns=["coordonnees_geo", "geometry"], inplace=True)

    return df_processed


df_processed = preprocess_df(df)
display(df_processed.head())
```

```python
def plot_sites_location(df: pd.DataFrame) -> go.Figure:
    """
    Creates a Plotly scatter map to display the locations of counting sites.
    """
    fig = go.Figure(
        go.Scattermap(
            lat=df["lat"],
            lon=df["lon"],
            mode="markers+text",
            marker=dict(size=14, symbol="marker"),
            text=df["label"] + " (" + df["id_site"] + ")",
            textposition="top right",
            textfont=dict(size=12, color="black"),
        )
    )

    # Configure map layout
    fig.update_layout(
        map=dict(
            center=dict(lat=df["lat"].mean(), lon=df["lon"].mean()),
            zoom=13.75,
        ),
        showlegend=False,
        margin=dict(l=30, r=30, t=50, b=30),
        title="Locations of Counting Sites",
        height=400,
        width=800,
    )

    return fig
```

# Hourly Bike and Scooter Count

```python
def plot_multimodal_site(df: pd.DataFrame, site_id: str) -> px.line:
    """
    Plots hourly multimodal counts at a given site
    """
    # 1) Filter to the specified site
    df_site = df[df["id_site"] == site_id].copy()

    # 2) Resample to an hourly grid for each transport mode
    df_resampled = (
        df_site.set_index("t_paris")
        .groupby("mode")["nb_usagers"]
        .resample("h")
        .sum()
        .reset_index()
    )

    # 3) Create the line plot
    fig = px.line(
        df_resampled,
        x="t_paris",
        y="nb_usagers",
        color="mode",
        title=f"Multimodal Hourly Count – Site {site_id}",
        labels={
            "t_paris": "Date and Time",
            "nb_usagers": "Number of Users",
            "mode": "Transport Mode",
        },
    )

    return fig
```

## Aggregated counts + average profiles

```python
def plot_aggregated_counts(df: pd.DataFrame) -> go.Figure:
    """
    Creates an interactive Plotly figure with buttons to display
    aggregated user counts at hourly, daily, and monthly levels.
    """
    # Aggregate by frequency
    df_hour = df["nb_usagers"].resample("h").sum().reset_index()
    df_day = df["nb_usagers"].resample("d").sum().reset_index()
    df_month = df["nb_usagers"].resample("MS").sum().reset_index()

    # Traces
    trace_hour = go.Scatter(
        x=df_hour["t_paris"], y=df_hour["nb_usagers"], name="Hourly", visible=True
    )
    trace_day = go.Scatter(
        x=df_day["t_paris"], y=df_day["nb_usagers"], name="Daily", visible=False
    )
    trace_month = go.Scatter(
        x=df_month["t_paris"], y=df_month["nb_usagers"], name="Monthly", visible=False
    )

    # Buttons
    buttons = [
        dict(
            label="Hourly",
            method="update",
            args=[
                {"visible": [True, False, False]},
                {
                    "title": "Hourly Counts",
                    "xaxis": {"title": "Date / Time"},
                },
            ],
        ),
        dict(
            label="Daily",
            method="update",
            args=[
                {"visible": [False, True, False]},
                {
                    "title": "Daily Counts",
                    "xaxis": {"title": "Date"},
                },
            ],
        ),
        dict(
            label="Monthly",
            method="update",
            args=[
                {"visible": [False, False, True]},
                {
                    "title": "Monthly Counts",
                    "xaxis": {"title": "Month"},
                },
            ],
        ),
    ]

    fig = go.Figure(data=[trace_hour, trace_day, trace_month])
    fig.update_layout(
        updatemenus=[
            dict(
                type="buttons",
                direction="right",
                x=0.5,
                y=1.15,
                showactive=True,
                buttons=buttons,
            )
        ],
        title="Hourly Counts",
        yaxis_title="Number of Users",
        xaxis=dict(type="date", title="Date / Time"),
    )

    return fig


plot_aggregated_counts(df_scope).show()
```

```python
def plot_daily_profiles_by_month(df):
    """Creates a Plotly figure showing daily usage profiles across months."""
    # labels and corresponding month numbers
    months = [
        "Jan",
        "Feb",
        "Mar",
        "Apr",
        "May",
        "Jun",
        "Jul",
        "Aug",
        "Sep",
        "Oct",
        "Nov",
        "Dec",
    ]
    month_nums = list(range(1, len(months) + 1))

    # prepare auxiliary columns
    df2 = df.copy()
    df2["month"] = df2.index.month
    df2["date_only"] = df2.index.normalize()  # drop time for grouping

    # create 4×3 subplots, shared y-axis
    fig = make_subplots(
        rows=4,
        cols=3,
        subplot_titles=months,
        shared_yaxes=True,
        horizontal_spacing=0.05,
        vertical_spacing=0.10,
    )

    for idx, (m_num, m_label) in enumerate(zip(month_nums, months)):
        row = idx // 3 + 1
        col = idx % 3 + 1

        # filter for this month
        dfm = df2[df2["month"] == m_num]

        # plot each day in this month
        for date, grp in dfm.groupby("date_only"):
            hours = grp.index.hour
            counts = grp["nb_usagers"]
            fig.add_trace(
                go.Scatter(
                    x=hours,
                    y=counts,
                    mode="lines",
                    line=dict(color="rgba(31,119,180,0.2)"),
                    showlegend=False,
                    customdata=grp.index,  # original timestamps
                    hovertemplate=(
                        "Users: %{y:.0f}<br>"
                        "Date/Time: %{customdata|%Y-%m-%d %H:%M}<extra></extra>"
                    ),
                ),
                row=row,
                col=col,
            )

        # plot the hourly average for the month
        if not dfm.empty:
            dfm2 = dfm.copy()
            dfm2["hour"] = dfm2.index.hour
            mean_by_hour = dfm2.groupby("hour")["nb_usagers"].mean()
            fig.add_trace(
                go.Scatter(
                    x=mean_by_hour.index,
                    y=mean_by_hour.values,
                    mode="lines",
                    line=dict(color="blue"),
                    name="Average",
                    showlegend=(idx == 0),
                    hovertemplate=(
                        "Hour of day: %{x}h<br>Average: %{y:.1f}<extra></extra>"
                    ),
                ),
                row=row,
                col=col,
            )

        # ticks every 8 hours
        fig.update_xaxes(
            tickmode="array", tickvals=list(range(0, 24, 8)), row=row, col=col
        )
        # hide x-labels in the top row if desired
        if m_label in ("Jan", "Feb", "Mar"):
            fig.update_xaxes(showticklabels=False, row=row, col=col)

    # common layout tweaks
    fig.update_layout(
        title="Daily traffic profile by month",
        height=800,
        margin=dict(t=80, b=80, l=80, r=40),
    )

    # shared axis labels via annotations
    fig.add_annotation(
        dict(
            x=0.5,
            y=-0.15,
            text="Hour of day",
            showarrow=False,
            xref="paper",
            yref="paper",
        )
    )
    fig.add_annotation(
        dict(
            x=-0.1,
            y=0.5,
            text="Number of users",
            showarrow=False,
            textangle=-90,
            xref="paper",
            yref="paper",
        )
    )

    return fig


plot_daily_profiles_by_month(df_scope).show()
```

```python
def plot_weekly_profiles(df) -> Figure:
    """Creates a Plotly figure showing weekly traffic profiles (Monday–Sunday)."""
    # Prepare a copy and compute week identifiers
    df2 = df.copy()
    # Week period ending Sunday (so the period covers Mon–Sun)
    df2["week_id"] = df2.index.to_period("W-SUN").astype(str)
    # Compute position in the week: 0 = Monday 00:00, …, 167 = Sunday 23:00
    df2["hour_of_week"] = df2.index.dayofweek * 24 + df2.index.hour

    fig = Figure()

    # Plot each week's series with low opacity
    for _, grp in df2.groupby("week_id"):
        fig.add_trace(
            Scatter(
                x=grp["hour_of_week"],
                y=grp["nb_usagers"],
                mode="lines",
                line=dict(color="rgba(31,119,180,0.1)"),
                showlegend=False,
                # Pass the datetime index into customdata for hover formatting
                customdata=grp.index,
                hovertemplate=(
                    "Users: %{y:.0f}<br>"
                    "Date/Time: %{customdata|%Y-%m-%d %H:%M}<extra></extra>"
                ),
            )
        )

    # Compute positional average, keeping only full weeks of 168 points
    weekly_frames = [
        grp[["hour_of_week", "nb_usagers"]].set_index("hour_of_week")["nb_usagers"]
        for _, grp in df2.groupby("week_id")
        if len(grp) == 168
    ]
    if weekly_frames:
        df_w = pd.concat(weekly_frames, axis=1)
        mean_ser = df_w.mean(axis=1).sort_index()
        fig.add_trace(
            Scatter(
                x=mean_ser.index,
                y=mean_ser.values,
                mode="lines",
                line=dict(color="blue"),
                name="Average",
                hovertemplate=(
                    "Hour of cycle: %{x}h<br>Average: %{y:.1f}<extra></extra>"
                ),
            )
        )

    # Set ticks every 24 hours with weekday labels
    DAYS = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
    fig.update_xaxes(
        tickmode="array",
        tickvals=list(range(0, 168, 24)),
        ticktext=DAYS,
        title_text="Day of week",
    )
    fig.update_yaxes(title_text="Number of users")

    fig.update_layout(
        title="Weekly traffic profile (by day of week)",
        height=500,
        legend=dict(yanchor="top", y=0.99, xanchor="right", x=0.99),
        margin=dict(l=60, r=40, t=60, b=60),
    )

    return fig


# Usage
plot_weekly_profiles(df_scope).show()
```

## Highlight missing data by month

```python
def summarize_missing_by_month(df_scope: pd.DataFrame) -> pd.DataFrame:
    """
    Compute, for each past calendar month (excluding the current month),
    the number of missing hourly slots, total theoretical hours, and
    percentage missing—only for months with at least one missing hour.
    """
    # 1) Count missing and total hours per month (MS = month start)
    missing = df_scope["nb_usagers"].isna().resample("MS").sum()
    total = df_scope["nb_usagers"].resample("MS").size()

    summary = pd.DataFrame({"missing_hours": missing, "total_hours": total})
    summary["pct_missing"] = summary["missing_hours"] / summary["total_hours"] * 100

    # 2) Keep only months with at least one missing hour
    summary = summary[summary["missing_hours"] > 0]

    # 3) Exclude the current month
    current_month = pd.Timestamp.now().to_period("M")
    # Convert summary index (DatetimeIndex at MS) to PeriodIndex
    period_index = summary.index.tz_localize(None).to_period("M")
    # Filter out current month
    summary = summary[period_index != current_month]

    # 4) Reassign clean PeriodIndex
    summary.index = period_index[period_index != current_month]
    summary.index.name = "month"

    return summary
```

## Impute missing data

```python
def impute_weekly_hourly(df, start=None, end=None):
    """
    Impute missing hourly values in 'nb_usagers' only within [start, end]
    using the historical average for the same weekday & hour, without leaking future data.
    Returns the full DataFrame with an 'is_imputed' column.
    """
    filled = df.copy().sort_index()
    filled["is_imputed"] = 0

    idx = filled.index
    tz = idx.tz

    # Build inclusive start_ts / end_ts
    if start is not None:
        start_ts = pd.to_datetime(start)
        if tz is not None and start_ts.tzinfo is None:
            start_ts = start_ts.tz_localize(tz)
    else:
        start_ts = idx[0]

    # End: if no time specified (i.e. midnight)
    if end is not None:
        end_ts = pd.to_datetime(end)
        # bump to 23:00 if time()==00:00
        if end_ts.time() == time(0, 0):
            end_ts = end_ts + timedelta(hours=23)
        if tz is not None and end_ts.tzinfo is None:
            end_ts = end_ts.tz_localize(tz)
    else:
        end_ts = idx[-1]

    # Mask of points within [start_ts, end_ts]
    window = (idx >= start_ts) & (idx <= end_ts)

    # Only impute NaNs inside that window
    to_impute = idx[filled["nb_usagers"].isna() & window]

    for ts in to_impute:
        history = filled["nb_usagers"][:ts].dropna()
        if history.empty:
            continue
        dow, hr = ts.dayofweek, ts.hour
        sel = history[(history.index.dayofweek == dow) & (history.index.hour == hr)]
        if not sel.empty:
            filled.at[ts, "nb_usagers"] = sel.mean()
            filled.at[ts, "is_imputed"] = 1

    return filled


# Impute
end = "2024-08-31"
df_imputed = impute_weekly_hourly(df_scope, end=end)

# Now assert no NaNs remains
assert df_imputed.loc[:end, "nb_usagers"].isna().sum() == 0
```

```python
def plot_hourly_with_imputed(df: pd.DataFrame) -> go.Figure:
    """Create a Plotly figure showing hourly counts with imputed values highlighted."""
    # Full series as a line
    fig = go.Figure()
    fig.add_trace(
        go.Scatter(
            x=df.index,
            y=df["nb_usagers"],
            mode="lines",
            name="Observed",
            line=dict(width=2),
        )
    )

    # Imputed points as red markers
    imputed = df[df["is_imputed"] == 1]
    fig.add_trace(
        go.Scatter(
            x=imputed.index,
            y=imputed["nb_usagers"],
            mode="markers",
            name="Imputed",
            marker=dict(color="red", size=6),
        )
    )

    # Layout
    fig.update_layout(
        title="Hourly Counts with Imputed Values Highlighted",
        xaxis_title="Date / Time",
        yaxis_title="Number of Users",
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1),
        margin=dict(l=40, r=40, t=80, b=40),
        height=400,
    )
    return fig


# df_imputed from previous step
fig = plot_hourly_with_imputed(df_imputed)
fig.show()
```

# MSTL decomposition

```python
# Select the series
s = df_imputed["nb_usagers"].loc["2023-01-01":"2023-12-31"]

# Instantiate and fit MSTL
mstl = MSTL(
    s,
    periods=[24, 24 * 7],  # daily and weekly
)
res = mstl.fit()
```

```python
# Extract components
obs = res.observed
trend = res.trend
seasonal = res.seasonal  # DataFrame with seasonal_24 & seasonal_168
resid = res.resid

# Plot with Plotly
fig = make_subplots(
    rows=5,
    cols=1,
    shared_xaxes=True,
    vertical_spacing=0.02,
    subplot_titles=[
        "Observed",
        "Trend",
        "Daily Seasonality (24h)",
        "Weekly Seasonality (168h)",
        "Residual",
    ],
)

fig.add_trace(go.Scatter(x=obs.index, y=obs, name="Observed"), row=1, col=1)
fig.add_trace(go.Scatter(x=trend.index, y=trend, name="Trend"), row=2, col=1)
fig.add_trace(
    go.Scatter(x=seasonal.index, y=seasonal["seasonal_24"], name="Daily"), row=3, col=1
)
fig.add_trace(
    go.Scatter(x=seasonal.index, y=seasonal["seasonal_168"], name="Weekly"),
    row=4,
    col=1,
)
# Residual as scatter markers
fig.add_trace(
    go.Scatter(
        x=resid.index,
        y=resid,
        mode="markers",
        marker=dict(symbol="circle", size=4),
        name="Residual",
    ),
    row=5,
    col=1,
)

fig.update_layout(height=800, title_text="MSTL Decomposition (2023)", showlegend=False)

fig.show()
```
